## Task 1: A list of dictionaries

```python
# Given data
transactions = [
    {"id": 1, "product": "Laptop", "price": 1200, "quantity": 2, "region": "EU"},
    {"id": 2, "product": "Phone", "price": 800, "quantity": -1, "region": "US"},
    {"id": 3, "product": "Tablet", "price": 0, "quantity": 5, "region": "EU"},
    {"id": 4, "product": "Monitor", "price": 350, "quantity": 3, "region": "US"},
    {"id": 5, "product": "Laptop", "price": 1200, "quantity": 1, "region": "EU"},
    {"id": 6, "product": "Keyboard", "price": 75, "quantity": 10, "region": "APAC"},
]
```

```python
# a) Filter invalid transactions (price > 0 and quantity > 0)
valid_transactions = [
    t for t in transactions
    if t["price"] > 0 and t["quantity"] > 0
]

valid_transactions
```

```python
# b) Add a total field
for t in valid_transactions:
    t["total"] = t["price"] * t["quantity"]

valid_transactions
```

```python
# c) Compute revenue per region
revenue_by_region = {}

for t in valid_transactions:
    region = t["region"]
    revenue_by_region[region] = revenue_by_region.get(region, 0) + t["total"]

revenue_by_region
```

```python
# d) Region with highest revenue
max_region = max(revenue_by_region, key=revenue_by_region.get)
max_value = revenue_by_region[max_region]

print(f"The region with the highest total revenue is {max_region} with revenue {max_value}.")
```

## Task 2: Creating a function

```python
sales = [1200, 980, 1540, 870, 2010, 1650, 1300, 1100, 950, 1780, 2200, 1430, 1600, 1900]
```

```python
def summarise_sales(sales):
    # Total sales
    total = sum(sales)
    
    # Average sales (rounded to 2 decimals)
    average = round(total / len(sales), 2)
    
    # Best and worst day (index starting at 1)
    best_day = sales.index(max(sales)) + 1
    worst_day = sales.index(min(sales)) + 1
    
    # Days above average
    above_average_count = sum(1 for s in sales if s > average)
    
    # Return the results as a dictionary
    return {
        "total": total,
        "average": average,
        "best_day": best_day,
        "worst_day": worst_day,
        "above_average_count": above_average_count
    }
```

```python
result = summarise_sales(sales)
print(result)
```

## Task 3: Loops

```python
students = [
    {"name": "Anna", "grade": 85, "attendance": 0.92},
    {"name": "Erik", "grade": 58, "attendance": 0.75},
    {"name": "Mia", "grade": 91, "attendance": 0.98},
    {"name": "Leo", "grade": 44, "attendance": 0.60},
    {"name": "Sara", "grade": 72, "attendance": 0.88},
    {"name": "Tom", "grade": 67, "attendance": 0.70},
    {"name": "Nina", "grade": 95, "attendance": 0.99},
    {"name": "Olli", "grade": 50, "attendance": 0.65},
]
```

```python
highest_student = None
highest_grade = -1  # start with something very low

for student in students:
    if student["grade"] > highest_grade:
        highest_grade = student["grade"]
        highest_student = student["name"]

print(f"Highest grade (loop): {highest_student} with {highest_grade}")
```

## Task 4: Class and Methods

```python
class Portfolio:
    def __init__(self, owner):
        self.owner = owner
        self.holdings = {}  # ticker -> {shares, price_per_share}

    # a) Add or update asset
    def add_asset(self, ticker, shares, price_per_share):
        if ticker in self.holdings:
            self.holdings[ticker]["shares"] += shares
            self.holdings[ticker]["price_per_share"] = price_per_share
        else:
            self.holdings[ticker] = {
                "shares": shares,
                "price_per_share": price_per_share
            }

    # b) Total value of portfolio
    def total_value(self):
        value = 0
        for ticker, data in self.holdings.items():
            value += data["shares"] * data["price_per_share"]
        return value

    # c) Most valuable asset
    def most_valuable(self):
        highest_value = -1
        most_valuable_ticker = None

        for ticker, data in self.holdings.items():
            value = data["shares"] * data["price_per_share"]
            if value > highest_value:
                highest_value = value
                most_valuable_ticker = ticker

        return most_valuable_ticker

    # d) Summary string
    def __str__(self):
        return (f"Portfolio of {self.owner} | "
                f"Assets: {len(self.holdings)} | "
                f"Total Value: €{self.total_value():,.2f}")
```

```python
# Create portfolio
p = Portfolio("Elmer")

# Add 3 assets
p.add_asset("AAPL", 10, 180)
p.add_asset("TSLA", 5, 200)
p.add_asset("MSFT", 7, 320)

# Demonstrate methods
print("Total value:", p.total_value())
print("Most valuable:", p.most_valuable())
print(p)   # calls __str__()
```

## Task 5: Importing a module

```python
orders = [
    ("2025-01-15", "Laptop"), ("2025-01-22", "Phone"),
    ("2025-02-03", "Laptop"), ("2025-02-14", "Tablet"),
    ("2025-03-01", "Phone"), ("2025-03-10", "Phone"),
    ("2025-03-22", "Laptop"), ("2025-04-05", "Monitor"),
]
```

```python
from datetime import datetime

# Convert each date string to a datetime.date object
parsed_dates = []

for date_str, product in orders:
    date_obj = datetime.strptime(date_str, "%Y-%m-%d").date()
    parsed_dates.append(date_obj)

# Find earliest and latest dates
earliest = min(parsed_dates)
latest = max(parsed_dates)

# Compute number of days between them
difference = (latest - earliest).days

print(f"Earliest order: {earliest}")
print(f"Latest order: {latest}")
print(f"Number of days between orders: {difference}")
```

## Task 6: Array Operations and Broadcasting

```python
import numpy as np

# Rows: years (2022, 2023, 2024)
# Columns: quarters (Q1, Q2, Q3, Q4)
revenue = np.array([
    [210, 230, 190, 250],
    [220, 260, 200, 270],
    [250, 290, 230, 310]
])
```

```python
# a) Total annual revenue per year
annual_totals = revenue.sum(axis=1)
print("Annual totals:", annual_totals)
```

```python
# b) Mean quarterly revenue
quarterly_means = revenue.mean(axis=0)
print("Quarterly means:", quarterly_means)
```

```python
# c) Broadcasting adjustment for 2024
growth = np.array([1.05, 1.03, 1.08, 1.02])
adjusted_2024 = revenue[2] * growth
print("Adjusted 2024 revenue:", adjusted_2024)
```

```python
# d) Values greater than 250
high_values = revenue[revenue > 250]
print("Values exceeding 250:", high_values)
```

## Task 8: More Pandas

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(seed=123)
days = pd.date_range(start="2025-01-06", periods=28, freq="D")
visitors = rng.integers(200, 600, size=28)
conversion_rate = rng.uniform(0.02, 0.08, size=28).round(4)
avg_order_value = rng.normal(50, 10, size=28).round(2)
```

```python
# a) Create DataFrame and set date as index
df = pd.DataFrame({
    "date": days,
    "visitors": visitors,
    "conversion_rate": conversion_rate,
    "avg_order_value": avg_order_value
})

df = df.set_index("date")
df
```

```python
# b) Compute conversions and daily revenue
df["conversions"] = (df["visitors"] * df["conversion_rate"]).round().astype(int)
df["daily_revenue"] = df["conversions"] * df["avg_order_value"]

df
```
